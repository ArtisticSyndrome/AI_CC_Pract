```python
def bfs(start, graph):
    visited = []
    queue = []
    queue.append(start)
    visited.append(start)
    while queue:
        ele = queue.pop(0)
        print(ele)
        for element in graph[ele]:
            if element not in visited:
                queue.append(element)
                visited.append(element)
```

```python
visited2 = []
def dfs(start,graph, visited2):
    stack = []
    stack.append(start)
    visited2.append(start)

    while stack:
        ele = stack.pop()
        print(ele, end='-->')
        for i in range(len(graph[ele])-1,-1,-1):
            neighbour = graph[ele][i]
            if neighbour not in visited2:
                stack.append(neighbour)
                visited2.append(neighbour)
```

```python
dfs('2',graph,visited2)
```

```python
visited3 = []
queue = []

def rec_bfs(node, graph, queue, visited3):
    if node not in visited3:
        queue.append(node)
        visited3.append(node)
    if not queue:
        return 
    
    ele = queue.pop(0)
    print(ele, end = '-->')
    
    for element in graph[ele]:
        if element not in visited3:
            queue.append(element)
            visited3.append(element)
    if queue:
        rec_bfs(queue[0], graph, queue, visited3)
```

```python
rec_bfs('2', graph, queue, visited3)
```

```python
def selection_sort(lst):
    for i in range(len(lst)):
        min = i
        for j in range(i+1,len(lst)):
            if lst[j]< lst[min]:
                min = j
        lst[i], lst[min] = lst[min], lst[i]
    return lst
```

```python
selection_sort([2,45,0,11,9])
```

```python
colors = ['red', 'blue','green']

states = ["WA", "NT", "SA", "Q", "NSW", "V"]

neighbors = { 
    "WA" : ["NT", "SA"],
    "NT" : ["WA", "SA", "Q"],
    "SA" : ["WA", "NT", "Q", "NSW", "V"],
    "Q" : ["NT", "SA", "NSW"],
    "NSW" : ["Q", "SA", "V"],
    "V" : ["SA", "NSW"]
}
color_state = {}
```

```python
def promising(node, color):
    for neighbour in neighbors.get(node):
        n_color = color_state.get(neighbour)
        if n_color == color:
            return False
    return True

def get_state_color(state):
    for color in colors:
        if promising(state, color):
            return color
    
```

```python
for state in states:
    color_state[state] = get_state_color(state)
color_state
```

```python
class Symptom_checker:
    def __init__(self):
        self.symptoms = {'cold' : False,
                    'fever' : False,
                    'headache' : False}
    def ask_symptoms(self):
        for s in self.symptoms:
            ans = input(f'Do you have {s} ?')
            if ans == 'yes':
                self.symptoms[s] = True

    def diagnose(self):
        if self.symptoms['cold'] & self.symptoms['fever']:
            print("you may have flu")
```

```python
obj = Symptom_checker()
obj.ask_symptoms()
obj.diagnose()
```
